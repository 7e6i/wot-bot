```python
import redis
import time

# Connect to Redis
r = redis.Redis(host='localhost', port=6379)
```

```python
print("Sending capture START request...")
r.xadd('record:request', {
    'command': 'start',
    'fps': 60,
    'clip_type': 'default'
})
```
